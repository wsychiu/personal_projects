```python
from IPython.display import display, HTML
import os
import pickle
import json
import time

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import re
import spacy
from spacy_cleaner import processing, Cleaner

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import (
    StratifiedKFold,
    GridSearchCV,
    cross_validate
) 
from sklearn.base import clone
from sklearn.metrics import (
    get_scorer,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    precision_recall_fscore_support
)

from openai import OpenAI

from utils import TextProcessing

# File locations
dirpath_data = 'datasets/'
fpath_fake_data = dirpath_data + 'Fake.csv'
fpath_real_data = dirpath_data + 'True.csv'
dirpath_models = 'models/'
dirpath_results = 'results/'

# RNG setup
random_state = 13
np.random.seed(random_state)
rng = np.random.default_rng()

# Loading spacy model and cleaner
spacy_model = spacy.load('en_core_web_sm')
# Cleaner removes stopwords and punctuations
#  and performs lemmatization (preferred over stemming)
text_cleaner = Cleaner(
    spacy_model,
    processing.remove_stopword_token,
    processing.remove_punctuation_token,
    processing.mutate_lemma_token
)
```

```python
# Basic dataset description and cleaning
news_data_fake = pd.read_csv(fpath_fake_data)
news_data_real = pd.read_csv(fpath_real_data)

def describe_df(df, data_title: str):
    print(f'{data_title} ---- ')
    print(f' > shape: {df.shape}')
    print(f' > duplicated records: {df.duplicated().sum()}')
    print(f' > empty records: {df.isnull().all(axis=1).sum()}')
   
    return pd.DataFrame({
        'Example' : df.iloc[0],
        'Data Type' : df.dtypes,
        'NaN Values' : df.isnull().sum(),
        'NaN %' : np.round(df.isnull().mean()*100, 0),
        'Unique Records' : df.nunique(),
        'Duplicated Records' : df.shape[0] - df.nunique()
    })

display(describe_df(news_data_fake, 'Fake News Data'))
print()
display(describe_df(news_data_real, 'Real News Data'))
```

```python
# Basic data cleaning - removing empty records and duplicated records
def basic_data_clean(df):
    df['date'] = pd.to_datetime(df['date'].str.strip(), format='%B %d, %Y')
    return df.dropna(how='all').drop_duplicates().copy()

news_data_fake_cleaned = basic_data_clean(news_data_fake)
news_data_real_cleaned = basic_data_clean(news_data_real)

display(describe_df(news_data_fake_cleaned, 'Fake News (cleaned)'))
display(describe_df(news_data_real_cleaned, 'Real News (cleaned)'))
```

```python
# Both datasets seem to share the same features - okay to join
# Selecting Fake News = 1 (target feature) and Not-Fake News = 0
news_data_fake_cleaned['target'] = 1
news_data_real_cleaned['target'] = 0

news_data_joined = pd.concat([news_data_fake_cleaned, news_data_real_cleaned])
display(describe_df(news_data_joined, 'Joined Data'))
news_data_joined.drop(columns='target').to_csv(dirpath_data + 'joined_data.csv')
```

```python
# Train-Test Split
# Total dataset is ~25K rows - not an overwhelming abundance of training data
# Suggest using 80-20 train-test split, with validation using Cross Validation
# Stratified, as proportion of fake news is low and we want to ensure training
#   is conducted with some samples of fake news
# Keeping target with data to handle potential row duplication during EDA

df_train, df_test, _, __ = train_test_split(
    news_data_joined,
    news_data_joined['target'],
    train_size = 0.8,  # 80% train proportion due to limited data size
    shuffle=True,
    stratify=news_data_joined['target'],
)

print(f'Training data: {df_train.shape}')
print(f'Test data: {df_test.shape}')
```

```python
# EDA
fig, axs = plt.subplots(2,3, 
                        figsize=(12, 8))
fig.tight_layout()
fig.subplots_adjust(hspace=0.6)

df_eda = df_train.copy()
df_eda['target_label'] = df_train['target'].replace({0:'Real', 1:'Fake'})

target_labels = ['Real', 'Fake']
real_news_mask = df_eda['target']==0
fake_news_mask = df_eda['target']==1
masks = (real_news_mask, fake_news_mask)

target_colour = {
    'Real' : 'grey',
    'Fake' : 'red'
}

def format_subplot(ax, title: str, show_y=False):
    ax.set_title(title, fontweight='bold')
    ax.get_yaxis().set_visible(show_y)
    ax.margins(y=0.4)
    return None

# Class (im)-balance
ax1 = axs[0][0]
format_subplot(ax1, '# Articles by Label', show_y=False)

counts = df_eda['target'].value_counts().sort_index()
class_imbalance = ax1.bar(
    target_labels,
    counts
)
ax1.bar_label(
    class_imbalance,
    labels=['{:,}'.format(count)
        for count in counts],
    label_type = 'center'
)

# Distribution of subject vs target
ax2 = axs[0][1]
subject_counts = (df_eda
                  .groupby(['subject', 'target_label'])
                  .size()
                  .unstack())
subject_counts.plot(kind='bar', ax=ax2, color=target_colour)

format_subplot(ax2, '# Articles by Subject and Label', show_y=False)
ax2.legend(loc='upper center', ncols=len(target_labels))
ax2.tick_params(axis='x', labelrotation=0)

# Distribution of dates by target label
ax3 = axs[0][2]

date_group = df_eda.groupby('target_label')['date']
for (label, series) in date_group:
    series.hist(
        bins=50,
        label=label,
        alpha=0.5,
        color=target_colour[label],
        ax=ax3
    )

format_subplot(ax3, 'Date histogram by target label', show_y=False)
ax3.tick_params(axis='x', labelrotation=90)
ax3.legend(loc='upper left', ncols=len(target_labels))

# Distribution of title and text length by target label
for [ax,col] in [
    [axs[1][0],'title'],
    [axs[1][1],'text']
]:
    ax.boxplot(
        [
            df_eda[real_news_mask][col].str.len(),
            df_eda[fake_news_mask][col].str.len(),
        ],
        tick_labels = target_labels
    )
    format_subplot(ax, 'Boxplot: # Chars in "' + col + '"', show_y=True)

# Distribution of Tags and URLs
ax4 = axs[1][2]

re_pattern = '@\w+|#\w+'
re_pattern += '|https?://\S+|www\.\S+'

format_subplot(ax4, 'Boxplot: # Tags & URLs in "title"', show_y=True)
ax4.boxplot([
                df_eda[masks[i]]['text'].str.count(re_pattern)
                for i in np.arange(len(masks))
            ],
           tick_labels=target_labels)

plt.savefig(dirpath_results + 'eda_plots.jpg')
plt.show()
```

```python
# Data Pre-processing
def text_normalization(text):
    # Cleaning data to remove tags (@, #) and websites
    re_pattern = '@\w+|#\w+'
    re_pattern += '|https?://\S+|www\.\S+'
    normalized_text = text.str.replace(
        re_pattern, '', regex=True
    )
    
    # Removing extra whitespace
    normalized_text = normalized_text.str.replace(
        '\s+', ' ', regex=True
    )

    # Removing punctuation and stop words, lemmatizing
    processed_text = spacy_model.pipe(
        normalized_text,
        disable=['parser', 'ner'],  # don't need parses and named entity recog.
        batch_size=500
    )

    # Returning in lowercase
    return [
        ' '.join(
            token.lemma_.lower()
            for token in text
            if not token.is_stop and not token.is_punct
        )
        for text in processed_text
    ]

def data_preprocessing(df):
    # Removing subject to prevent target leakage
    df_cleaned = df.drop(columns='subject')    
        
    # Enforcing data  types
    df_cleaned = df_cleaned.astype({
        'title' : 'string',
        'text' : 'string',
        'date' : 'datetime64[ns]',
    })

    # Removing instances where text + title are duplicated
    df_cleaned.drop_duplicates(
        subset=['title', 'text'],
        keep='first',
        inplace=True
    )

    # Removing records where there is no text in article
    empty_text = df['text'].str.strip().str.len()==0
    df_cleaned.drop(df[empty_text].index, inplace=True)

    df_cleaned['normalized_title'] = text_normalization(df_cleaned['title'])
    df_cleaned['normalized_text'] = text_normalization(df_cleaned['text'])

    # Adding counts of @/# tags and URLs
    re_pattern = '@\w+|#\w+'
    re_pattern += '|https?://\S+|www\.\S+'
    df_cleaned['tag_url_count'] = df_cleaned['text'].str.count(re_pattern)
    
    return df_cleaned.drop(columns=['title', 'text'])

# Pre-processing data (prior to tf-idf vectorization and feature scaling)
df_train_cleaned = data_preprocessing(df_train[:1000])
df_test_cleaned = data_preprocessing(df_test[:1000])

# Separating target from training and test datasets
X_train = df_train_cleaned.drop(columns='target')
y_train = df_train_cleaned['target']

X_test = df_test_cleaned.drop(columns='target')
y_test = df_test_cleaned['target']

display(describe_df(df_train_cleaned, 'Pre-processed Joined Data'))
```

```python
# Model Pipeline
data_transformer = ColumnTransformer(
    transformers = [
        ('tfidf_title', TfidfVectorizer(), 'normalized_title'),
        ('tfidf_text', TfidfVectorizer(), 'normalized_text'),
        ('tag_url_count', StandardScaler(), ['tag_url_count']),
    ],
    remainder='drop'
)

lr_pipeline = Pipeline(
    [
        ('transform_data', data_transformer),
        ('logisticregression', LogisticRegression(class_weight='balanced'))
    ]
)

# Cross-validation
inner_cv = StratifiedKFold(
    n_splits = 5,
    shuffle = True,
    random_state = random_state
)

outer_cv = StratifiedKFold(
    n_splits = 5,
    shuffle = True,
    random_state = random_state
)

param_grid = [
    {'logisticregression__C' : [0.001, 0.01, 0.1, 1, 10, 100, 1000],
     'logisticregression__solver' : ['lbfgs', 'liblinear'],
    }
]
```

```python
gridsearch_lr = GridSearchCV(
    estimator = lr_pipeline,
    param_grid = param_grid,
    scoring = 'f1_macro',
    cv = inner_cv
)

gridsearch_lr.fit(X_train, y_train)
```

```python
# Determining performance and threshold for overfitting
lr_model = gridsearch_lr.best_estimator_
df_results = pd.DataFrame(gridsearch_lr.cv_results_)

for solver, group in df_results.groupby('param_logisticregression__solver'):
    group = group.sort_values('param_logisticregression__C')
    plt.errorbar(
        group['param_logisticregression__C'],
        group['mean_test_score'],
        yerr = group['std_test_score'],
        marker = 'o',
        capsize = 3,
        label = solver,
        alpha = 0.7
    )

plt.xscale('log')
plt.xlabel('log(C)')
plt.ylabel('F1 Score')
plt.title('Regularization Strength vs F1 Score', fontweight='bold')
plt.legend(loc='upper left')

plt.savefig(dirpath_results + 'lr_regularization_vs_f1.jpg')
plt.show()
```

```python
# Choosing 'liblinear' and C = 10 as that seems to be the inflection
#   point for training loss plateau, indicating start of overfitting
lr_model = clone(gridsearch_lr.estimator)
lr_model.set_params(
    logisticregression__C = 1,
    logisticregression__solver = 'liblinear'
)
lr_model.fit(X_train, y_train)
```

```python
scores = [
    'precision',
    'recall',
    'f1',
    'accuracy'
]

cv_scores = cross_validate(
    lr_model,
    X_train,
    y_train,
    cv = outer_cv,
    scoring = scores,
    return_train_score=True
)
display(pd.DataFrame(cv_scores).round(3))
# Model seems not to vary wildly between train and test
#  indicating good generalization performance
```

```python
# Classifier Performance Metrics
def classifier_scores(y_test, y_pred, labels):
    scores = {}
    scores['accuracy'] = accuracy_score(y_test, y_pred)
    scores['precision'], scores['recall'], scores['f1'], _ = precision_recall_fscore_support(
        y_test, y_pred
    )
    cm = ConfusionMatrixDisplay(
        confusion_matrix(y_test, y_pred),
        display_labels = labels
    )
    return scores, cm

lr_y_pred = lr_model.predict(X_test)
lr_scores, lr_cm = classifier_scores(y_test, lr_y_pred, ['Real', 'Fake'])
```

```python
# GenAI-based approach: Zero-shot prompting
llm_client = OpenAI()
llm_model = 'gpt-5.4-mini'

system_prompt = '''Your task is to produce an analysis of a news article and provide reasons why the news article is likely Real or Fake news.
You will be provided the Title and Text of a list of news article and you must provide your response in the following JSON list format:
[{"Article" : int, "Classification" : "REAL" or "FAKE"}]
'''
```

```python
# Batch generator
def batch_records(records, batch_size=20):
    for batch_num in range(0, len(records), batch_size):
        yield records.iloc[batch_num : batch_num+batch_size]
```

```python
def llm_batch_classify(model, llm_model, system_prompt, batch):
    article_list = ""
    for i in range(len(batch)):
        article_list += (
            f"Article: {i}.\n"
            f"Title: {batch.iloc[i]['title']}\n"
            f"Text: {batch.iloc[i]['text']}\n\n"
        )

    classify_response = model.responses.create(
        model = llm_model,
        input = [
            {
                'role' : 'system',
                'content' : (system_prompt)
            },
            {
                'role' : 'user',
                'content' : (article_list)
            }
        ],
        temperature=0
    )
    return classify_response
```

```python
def llm_classify(model, llm_model, system_prompt, batches):
    llm_classifications = pd.DataFrame()
    run_times = []

    for i, batch in enumerate(batches):
        batch_size = batch.shape[0]
        print(f'Classifying Batch: {i} ({batch_size} articles')
        
        start_time = time.perf_counter()
        classify_response = llm_batch_classify(llm_client, llm_model, system_prompt, batch)
        end_time = time.perf_count()
        run_times.append(end_time - start_time)
        
        y_pred = (
            pd.DataFrame(json.loads(classify_response.output_text))
             ['Classification']
            .map({'REAL':0, 'FAKE':1})
            .rename('llm_y_pred')
        )

        classifications = pd.concat([y_pred, batch['target'].reset_index(drop=True)], axis=1)
        llm_classifications = pd.concat([llm_classifications, classifications], axis=0)
    
    return llm_classifications.reset_index(drop=True), run_times
```

```python
batches = batch_records(df_train[:100])
llm_classifications, llm_times = llm_classify(llm_client, llm_model, system_prompt, batches)
```

```python
# Classifier performance metrics
y_test = llm_classifications['target']
llm_y_pred = llm_classifications['llm_y_pred']

llm_scores, llm_cm = classifier_scores(y_test, llm_y_pred, ['Real', 'Fake'])
```

```python
# Printing and saving scores
def score_to_df(scores):
    df = pd.DataFrame(scores).T
    df['Macro Mean'] = df.mean(axis=1)
    return df

df_scores = pd.concat(
    {
        "lr" : score_to_df(lr_scores),
        "llm" : score_to_df(llm_scores)
    },
    axis=1
)

display(df_scores.round(2))
df_scores.to_csv(
    dirpath_results + 'classifier_scores.csv',
    
)

# Plotting Confusion Matrices
fig, axs = plt.subplots(1,2, figsize=(12, 5))
for i, ax in enumerate(axs):
    cm = [lr_cm, llm_cm][i]
    cm.plot(ax=ax)
    ax.set_title(
        f'{["Logistic Regression", "GPT-5.4-Mini"][i]} classifier Confusion Matrix',
        fontweight='bold'
    )
    
plt.tight_layout()
plt.savefig(dirpath_results + 'confusion_matrices.jpg')
plt.show()
```
